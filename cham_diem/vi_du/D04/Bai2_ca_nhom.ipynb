# Bài 2 - Những con số dấu ấn

## Thành viên A

```
data = [5, 19, 8, 13, 7]
mx = mn = data[0]
for x in data:
    if x > mx: mx = x
    if x < mn: mn = x
print('A =', mx - mn)
```

## Thành viên B

```
data = [1, 2, 5, 16, 19]
lon = [None, None]; nho = [None, None]
for x in data:
    if lon[0] is None or x > lon[0]: lon = [x, lon[0]]
    elif lon[1] is None or x > lon[1]: lon[1] = x
    if nho[0] is None or x < nho[0]: nho = [x, nho[0]]
    elif nho[1] is None or x < nho[1]: nho[1] = x
print('B =', lon[1] - nho[1])
```

## Thành viên C

```
%time
data = [9, 1, 7, 1]
tong = 0
for i, x in enumerate(data):
    tong += x if i % 2 == 0 else -x
tong
```
